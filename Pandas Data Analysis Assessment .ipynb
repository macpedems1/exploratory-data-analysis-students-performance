# SECTION B

#### Selecting all students who completed the test preparation course

```python
prep_completed = df[df["test preparation course"] == "completed"]
prep_completed.head()
```

All students who scored above 70 in math

```python
math_above_70 = df[df["math score"] > 70]
math_above_70.head()
```

female students who scored at least 65 in all three subjects (using the & operator)

```python
female_high_achievers = df[
    (df["gender"] == "female")
    & (df["math score"] >= 65)
    & (df["reading score"] >= 65)
    & (df["writing score"] >= 65)
]
female_high_achievers.head()
```

Number of student with free/ reduced lunch

```python
free_reduced_count = (df["lunch"] == "free/reduced").sum()
print(f"\nNumber of students with free/reduced lunch: {free_reduced_count}")
```

# SECTION C - GROUPBY AND AGGREGATION

Average math score by gender

```python
avg_math_by_gender = df.groupby("gender")["math score"].mean()
print("\n--- Average Math Score by Gender ---")
print(avg_math_by_gender)
```

Average scores (math, reading, writing) by parental level of education. 

```python
avg_scores_by_parent_edu = df.groupby("parental level of education")[
    ["math score", "reading score", "writing score"]
].mean()

print(avg_scores_by_parent_edu)
```

 Comparing average scores of students with vs without test preparation

```python
avg_scores_by_prep = df.groupby("test preparation course")[
    ["math score", "reading score", "writing score"]
].mean()

print(avg_scores_by_prep)
```

# SECTION D - VISUALIZATION

 Score Histogram

```python
# 1. First, create the figure and axes objects!
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 2. Math Score Histogram
axes[0].hist(df["math score"], bins=15, color="skyblue", edgecolor="black")
axes[0].set_title("Distribution of Math Scores", fontsize=14, fontweight="bold")
axes[0].set_xlabel("Math Score", fontsize=12)
axes[0].set_ylabel("Number of Students", fontsize=12)

# 3. Reading Score Histogram
axes[1].hist(df["reading score"], bins=15, color="salmon", edgecolor="black")
axes[1].set_title("Distribution of Reading Scores", fontsize=14, fontweight="bold")
axes[1].set_xlabel("Reading Score", fontsize=12)
axes[1].set_ylabel("Number of Students", fontsize=12)

# 4. Writing Score Histogram
axes[2].hist(df["writing score"], bins=15, color="lightgreen", edgecolor="black")
axes[2].set_title("Distribution of Writing Scores", fontsize=14, fontweight="bold")
axes[2].set_xlabel("Writing Score", fontsize=12)
axes[2].set_ylabel("Number of Students", fontsize=12)

# 5. Display the plot
plt.tight_layout()
plt.show()
```

Bar plot showing average score by gender. 

```python
avg_by_gender_all = df.groupby("gender")[
    ["math score", "reading score", "writing score"]
].mean()

# Plot group bar chart
ax = avg_by_gender_all.plot(kind="bar", figsize=(10, 6), width=0.7)
plt.title("Average Subject Scores by Gender", fontsize=14, fontweight="bold")
plt.xlabel("Gender", fontsize=12)
plt.ylabel("Average Score", fontsize=12)
plt.xticks(rotation=0)
plt.ylim(0, 100)
plt.legend(title="Subject")
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Annotate values on top of bars
for p in ax.patches:
    ax.annotate(
        f"{p.get_height():.1f}",
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha="center",
        va="center",
        xytext=(0, 5),
        textcoords="offset points",
    )

plt.tight_layout()
plt.show()
```

# SECTION E - INTERPRETATION OF ANALYSIS


Yes, students who completed the test preparation course consistently performed better than those who didn't, across every subject. On average, completers scored about 69.7 in math versus 64.1 for non-completers, 73.9 versus 64.8 in reading, and 74.4 versus 64.3 in writing. Writing showed the biggest gain, with roughly a 10-point boost. 
Taken together, the data points to a clear, consistent link between structured test prep and stronger academic performance overall
